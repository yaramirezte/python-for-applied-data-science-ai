# Exception Handling

```python
def safe_divide(numerator, denominator ):
    try:
        c=numerator/denominator   
        return c
    except ZeroDivisionError:
        print("it's not possible to divide by zero")
        return None
a=int(input("Please enter the numerator:"))
b=int(input("Please enter the denominator:"))
print(safe_divide(a,b))
```

```python
import math
def safe_squareroot(number1):
    try:
        result=math.sqrt(number1)
        return result
    except ValueError:
        print("Invalid input! Please enter a positive integer or a float value")


N1=float(input("Please enter the number:"))
print(safe_squareroot(N1))
```

```python
def complex_calculation(num):
    try:
        result = num / (num - 5)
        print (f"Result: {result}")
    except Exception as e:
        print("An error occurred during calculation.")
# Test case
user_input = float(input("Enter a number: "))
complex_calculation(user_input)
```
